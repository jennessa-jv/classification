```python
import torch
from torch import nn, optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
#normalize data
tf = transforms.Compose([  #makes it act like a function
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
])
train_dl=DataLoader(
    datasets.ImageFolder('data/Training',tf),
    batch_size=32,shuffle=True,num_workers=4
)
test_dl=DataLoader(
    datasets.ImageFolder('data/Testing',tf),
    batch_size=32,shuffle=False,num_workers=4
)
```

```python
model = nn.Sequential(
    nn.Conv2d(3, 32, 3, 1, 1), nn.ReLU(), nn.MaxPool2d(2), #here the weightd get modified based on every epoch
    nn.Conv2d(32, 64, 3, 1, 1), nn.ReLU(), nn.MaxPool2d(2),
    nn.Conv2d(64, 128, 3, 1, 1), nn.ReLU(), nn.MaxPool2d(2),
    nn.Flatten(),
    nn.Linear(128 * 16 * 16, 256), nn.ReLU(), nn.Dropout(0.5),
    nn.Linear(256, 4)
).to(device)
```

```python
opt = optim.AdamW(model.parameters(), 1e-4) #2nd argument is the learning rate
loss_fn = nn.CrossEntropyLoss()   #loss_fn = nn.CrossEntropyLoss(reduction='sum') --if i wanted to see the sum per batch
```

```python
model.train()

for epoch in range(25):
    running_loss = 0

    for x, y in train_dl:
        opt.zero_grad()

        predictions = model(x.to(device))
        actual_answers = y.to(device)

        loss = loss_fn(predictions, actual_answers)
        loss.backward() #here is where the individual gradients are calculated

        running_loss += loss.item()

        opt.step() #the gradients here adjust the weights

    print(f'Epoch {epoch+1}: Loss was {running_loss}')
```

```python
model.eval()
test_loss, correct = 0.0, 0

with torch.no_grad():
    for x, y in test_dl:
        x, y = x.to(device), y.to(device)

        logits = model(x)

        test_loss += loss_fn(logits, y).item() * y.size(0)

        preds = logits.argmax(dim=1)
        correct += (preds == y).sum().item()

test_loss /= len(test_dl.dataset)
accuracy = 100.0 * correct / len(test_dl.dataset)

print("Test Loss:", test_loss, "Test accuracy:", accuracy, "%")
```

```python
import random
import matplotlib.pyplot as plt
from torchvision.transforms.functional import to_pil_image

model.eval()

idx = random.randrange(len(test_dl.dataset))
img, label = test_dl.dataset[idx]

unnorm = img * 0.5 + 0.5
plt.imshow(to_pil_image(unnorm))
plt.axis('off')
plt.title('Sample from test set')
plt.show()

with torch.no_grad():
    logits = model(img.unsqueeze(0).to(device))
    pred = logits.argmax(1)

class_names = test_dl.dataset.classes
print(f"Predicted class: {class_names[pred]}")
print(f"Ground-truth: {class_names[label]}")
```
